## 1. Create variables of every kind
They appear in the workspace table on their own (it refreshes after each cell).

```python
import numpy as np, sympy as sp, scipy.sparse as sps, pandas as pd

rng = np.random.default_rng(0)
t = np.linspace(0, 10, 2000)
signal = np.exp(-0.15 * t) * np.sin(2 * np.pi * 1.2 * t) + 0.04 * rng.standard_normal(t.size)
spectrum = np.abs(np.fft.rfft(signal))
A = np.round(rng.normal(size=(8, 6)), 3)
cube = rng.random((4, 32, 32))                       # N-d: use the slice selector
X, Y = np.meshgrid(np.linspace(-3, 3, 140), np.linspace(-3, 3, 140))
psi = np.exp(-(X**2 + Y**2) / 2) * np.cos(3 * X)     # shown as an image
z = np.exp(1j * np.linspace(0, 6, 7))                # complex numbers
S = sps.random(2000, 2000, density=0.002, format="csr")   # SciPy sparse
x, a, b = sp.symbols("x a b")
f = sp.integrate(sp.sin(x)**2 * sp.exp(-x), x)       # SymPy -> LaTeX
H = sp.Matrix([[a, b], [b, -a]])
df = pd.DataFrame({"t": t[:200:25], "V": np.sin(t[:200:25])})
params = {"m": 1.0, "k": 4.0, "modes": [1, 2, 3]}
```

## 2. Try it
- **Double-click** `signal`, `psi`, `f`, `df`, `params`… each one opens in its own tab and **stays open**.
- `signal`: *Line*, *Histogram*, *Bars* tabs. `psi`: *Image*.
- Right-click the workspace header to add statistics columns (`Mean`, `Std`, `Max`…).
- `A`: double-click a cell to **edit** it, then print `A` below.
- `cube`: change the **slice** selector `[i, :, :]`.
