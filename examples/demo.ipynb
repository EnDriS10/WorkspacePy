## 1. Crea variables de todo tipo
Aparecen solas en el workspace (se refresca al terminar cada celda).

```python
import numpy as np, sympy as sp, scipy.sparse as sps, pandas as pd

t  = np.linspace(0, 10, 2000)
senal = np.sin(2*np.pi*0.5*t) + 0.2*np.random.default_rng(0).standard_normal(t.size)
A  = np.arange(12.).reshape(3, 4)
T  = np.random.default_rng(1).random((4, 5, 6))      # N-d: usa el selector de corte
img = np.add.outer(np.sin(np.linspace(0, 6, 80)), np.cos(np.linspace(0, 6, 120)))
z  = np.array([1+2j, 3-1j])
S  = sps.random(500, 500, density=0.01, format='csr')  # SciPy sparse
x  = sp.Symbol('x')
f  = sp.integrate(sp.sin(x)**2, x)                     # SymPy
M  = sp.Matrix([[1, x], [x**2, 2]])
df = pd.DataFrame({'t': t[:5], 'v': senal[:5]})
datos = {'a': 1, 'b': [1, 2, 3], 'M': np.eye(3)}
```
